# Install Jupyter

Install Jupyter using pip or conda. Run `pip install jupyter` in the terminal.

# Launch Jupyter Notebook

Launch the Jupyter notebook server by running `jupyter notebook` in the terminal.

# Start a New Notebook

Create a new notebook by clicking 'New' and selecting 'Python 3' in the Jupyter interface.
